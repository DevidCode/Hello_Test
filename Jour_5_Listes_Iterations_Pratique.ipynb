## 🧾 Rappel rapide

Une **liste** est une collection ordonnée de valeurs, qu'on écrit entre crochets `[]`.

```python
fruits = ["pomme", "banane", "kiwi"]
print(fruits[0])  # "pomme"
```

On peut :
- ajouter un élément avec `.append()`
- modifier un élément par son indice
- la parcourir avec une boucle `for`


## 🧪 Exercice 1 : Créer une liste

1. Crée une liste contenant 3 prénoms.

3. Affiche uniquement le **deuxième prénom**.


````python
prenoms = ["Alice", "Bob", "Charlie"]

# Affiche la liste
print (prenoms)
# Affiche le deuxième prénom
print(prenoms[1])
````

## 🧪 Exercice 3 : Afficher chaque prénom

Utilise une boucle `for` pour afficher chaque prénom de la liste.


````python
prenoms = ["Alice", "Bob", "Charlie"]
for i in prenoms:
  print (i)
````

````python
# @title
for nom in prenoms:
    print(nom)
````

## 🧪 Exercice 4 : Afficher en majuscules

Affiche chaque prénom **en majuscules** (sans modifier la liste d’origine).


````python
prenoms = ["Alice", "Bob", "Charlie"]
for i in prenoms:
  print(f"{i.upper()}")
  prenoms[i
          ] = i.upper()
print(prenoms)
````

````python
# @title
for nom in prenoms:
    print(nom.upper())
````

## 🧪 Exercice 5 : Ajouter des prénoms

Demande à l’utilisateur **3 prénoms** et ajoute-les à une liste vide avec `.append()`.

Puis affiche la liste finale.


````python
liste = []

# Demande 3 fois à l'utilisateur et ajoute à la liste
for i in range(1,4):
  liste.append(str(input("Donne moi un prenom stp")))
# Affiche la liste finale
print (liste)
````
